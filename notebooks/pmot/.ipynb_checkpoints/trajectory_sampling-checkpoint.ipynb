# Vector-only pMOT: configurable single-atom trajectory

This notebook launches one Rb-87 atom from the standard 15 mm shell at 17 m/s and plots its trajectory together with the pMOT beams at their physical scale. The default launch is

\[
\mathbf r_0=(15,0,0)\ {\rm mm},\qquad
\mathbf v_0=(-17,0,0)\ {\rm m\,s^{-1}}.
\]

All 18 traveling optical components are independently configurable: six cooling, six repump, and six 1529 nm trapping components. Every polarization label is defined in that component's propagation frame (the observer looks along its own \(\mathbf k\)). The centered restoring-and-damping configuration found by the current helicity study is selected by default:

- incident and retro cooling: \((\sigma^+,\sigma^+,\sigma^-)\) on \((x,y,z)\);
- incident and retro repump: \((\sigma^+,\sigma^+,\sigma^-)\);
- incident and retro trapping: \((\sigma^+,\sigma^+,\sigma^-)\).

Equal propagation-frame labels on counterpropagating components correspond to opposite laboratory-frame optical-spin vectors.


```python
import os
import sys
import json
from dataclasses import asdict, replace
from datetime import datetime
from pathlib import Path
from time import perf_counter

_backend = "inline" if os.environ.get("PMOT_NOTEBOOK_HEADLESS") == "1" else "widget"
get_ipython().run_line_magic("matplotlib", _backend)

import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display

PROJECT_ROOT = Path.cwd().resolve()
while PROJECT_ROOT != PROJECT_ROOT.parent:
    if (PROJECT_ROOT / "pyproject.toml").is_file() and (PROJECT_ROOT / "src" / "pmot").is_dir():
        break
    PROJECT_ROOT = PROJECT_ROOT.parent
else:
    raise RuntimeError("Could not locate the pMOT repository root.")

SRC_PATH = PROJECT_ROOT / "src"
if str(SRC_PATH) not in sys.path:
    sys.path.insert(0, str(SRC_PATH))

from pmot.mot_multilevel.rate_equations import RateEquationTrajectoryConfig
from pmot.pmot.trajectory_plotting import plot_pmot_trajectory_diagnostics
from pmot.pmot.vector_only_trajectories import (
    PMOTBeamHelicities,
    build_vector_only_apparatus,
    build_vector_only_trajectory_context,
    inward_launch_state,
    save_vector_only_trajectory_csv,
    simulate_vector_only_pmot_trajectory,
    vector_only_trajectory_dataframe,
)

DATA_OUTPUT_ROOT = PROJECT_ROOT / "outputs" / "trajectories" / "pmot" / "vector_only_single_atom"
FIGURE_OUTPUT_ROOT = PROJECT_ROOT / "outputs" / "figures" / "pmot" / "vector_only_single_atom"
print(f"Project root: {PROJECT_ROOT}")
print(f"Matplotlib backend: {_backend}")
```

## Physics boundary

This is the corrected, provisional **ideal-magic vector-only** pMOT model. The repumper-included 24-state rate equation for the 780 nm light is retained, while the external magnetic field is fixed to zero. At 1529.268881 nm the scalar and tensor differential shifts are imposed to cancel, and only the vector differential transition shift is applied:

\[
\Delta_{{\rm eff},eg}^{(j)}
=
\Delta_{0,j}-\mathbf k_j\!\cdot\!\mathbf v
-\frac{\Delta E_{eg}^{(v)}(\mathbf r,\mathbf v)}{\hbar}.
\]

The 1529 nm field changes the transition-dependent effective detunings used by the 780 nm absorption-rate calculation. The plotted force is the resulting 780 nm absorption-momentum force; gravity is integrated separately. This model does **not yet** include a direct mechanical 1529 nm force, conservative Stark-gradient force, or trapping-light scattering/heating, so its capture result remains provisional.


```python
AXES = ("x", "y", "z")
KNOWN_WORKING_XYZ = ("sigma+", "sigma+", "sigma-")
DEFAULT_HELICITIES = PMOTBeamHelicities(
    cooling_incident_xyz=KNOWN_WORKING_XYZ,
    cooling_retro_xyz=KNOWN_WORKING_XYZ,
    repump_incident_xyz=KNOWN_WORKING_XYZ,
    repump_retro_xyz=KNOWN_WORKING_XYZ,
    trapping_incident_xyz=KNOWN_WORKING_XYZ,
    trapping_retro_xyz=KNOWN_WORKING_XYZ,
)

def beam_manifest(context):
    rows = []
    for beam in context.cooling_repump_beams:
        rows.append({
            "family": beam.family,
            "axis": beam.axis_name,
            "path": beam.propagation_sense,
            "k_hat": tuple(round(value, 3) for value in beam.direction),
            "propagation-frame polarization": beam.circular_polarization,
            "wavelength [nm]": 1e9 * beam.wavelength_m,
            "power/component [mW]": 1e3 * beam.power_w,
            "1/e^2 diameter at waist [mm]": 2e3 * beam.beam_radius_m,
            "waist coordinate [mm]": np.nan,
        })
    for beam in context.trapping_beams:
        rows.append({
            "family": "trapping",
            "axis": beam.axis_name,
            "path": beam.propagation_sense,
            "k_hat": tuple(round(value, 3) for value in beam.direction),
            "propagation-frame polarization": beam.helicity,
            "wavelength [nm]": 1e9 * beam.wavelength_m,
            "power/component [mW]": 1e3 * beam.power_per_incident_path_watt,
            "1/e^2 diameter at waist [mm]": 2e3 * beam.waist_radius_m,
            "waist coordinate [mm]": 1e3 * np.dot(beam.waist_position_m, beam.direction),
        })
    return pd.DataFrame(rows)

def _context_with_gravity(context, include_gravity):
    if context.multilevel_config.include_gravity == include_gravity:
        return context
    config = replace(context.multilevel_config, include_gravity=bool(include_gravity))
    return replace(
        context,
        multilevel_config=config,
        _observable_context=replace(context._observable_context, multilevel_config=config),
    )

def _json_ready(value):
    if isinstance(value, dict):
        return {str(key): _json_ready(item) for key, item in value.items()}
    if isinstance(value, (list, tuple)):
        return [_json_ready(item) for item in value]
    if isinstance(value, np.ndarray):
        return value.tolist()
    if isinstance(value, (np.floating, np.integer, np.bool_)):
        return value.item()
    return value

def run_pmot_case(
    *,
    radial_distance_mm=15.0,
    speed_m_per_s=17.0,
    polar_angle_deg=90.0,
    azimuth_angle_deg=0.0,
    impact_parameter_mm=0.0,
    impact_azimuth_deg=0.0,
    duration_ms=25.0,
    time_step_us=5.0,
    include_diffusion=False,
    include_gravity=True,
    seed=20260901,
    escape_radius_mm=30.0,
    cooling_power_mw=27.0,
    repump_power_mw=0.1,
    cooling_detuning_mhz=-15.0,
    repump_detuning_mhz=0.0,
    beam_diameter_mm=12.7,
    trapping_wavelength_nm=1529.268881,
    focus_offset_mm=10.0,
    input_beam_diameter_mm=35.0,
    focal_length_mm=80.3,
    retro_power_fraction=1.0,
    derive_trapping_power=True,
    target_gradient_g_per_cm=20.0,
    explicit_trapping_power_mw=38.294486,
    helicities=DEFAULT_HELICITIES,
    plot_extent_mm=20.0,
    save_outputs=True,
    output_tag=None,
    progress_widget=None,
):
    apparatus = build_vector_only_apparatus(
        cooling_power_w_per_beam=1e-3 * cooling_power_mw,
        repump_power_w_per_beam=1e-3 * repump_power_mw,
        cooling_detuning_hz=1e6 * cooling_detuning_mhz,
        repump_detuning_hz=1e6 * repump_detuning_mhz,
        beam_diameter_m=1e-3 * beam_diameter_mm,
        trapping_wavelength_m=1e-9 * trapping_wavelength_nm,
        trapping_focus_offset_m=1e-3 * focus_offset_mm,
        trapping_input_beam_diameter_m=1e-3 * input_beam_diameter_mm,
        trapping_focal_length_m=1e-3 * focal_length_mm,
        trapping_retro_power_fraction=retro_power_fraction,
    )
    context = build_vector_only_trajectory_context(
        apparatus=apparatus,
        helicities=helicities,
        trapping_power_w_per_path=(
            None if derive_trapping_power else 1e-3 * explicit_trapping_power_mw
        ),
        target_gradient_g_per_cm=target_gradient_g_per_cm,
    )
    context = _context_with_gravity(context, include_gravity)
    launch = inward_launch_state(
        radial_distance_m=1e-3 * radial_distance_mm,
        speed_m_per_s=speed_m_per_s,
        polar_angle_deg=polar_angle_deg,
        azimuth_angle_deg=azimuth_angle_deg,
        impact_parameter_m=1e-3 * impact_parameter_mm,
        impact_azimuth_deg=impact_azimuth_deg,
    )
    numerical = RateEquationTrajectoryConfig(
        time_step_s=1e-6 * time_step_us,
        include_diffusion=include_diffusion,
        seed=int(seed),
        escape_radius_m=1e-3 * escape_radius_mm,
    )
    step_count = int(np.ceil(1e3 * duration_ms / time_step_us))
    print(
        f"Running {duration_ms:g} ms at dt={time_step_us:g} us "
        f"({step_count:,} steps), v0={speed_m_per_s:g} m/s; "
        f"trapping helicity code={context.trapping_code}",
        flush=True,
    )
    print(
        f"External B=(0,0,0); cooling={cooling_power_mw:g} mW/beam; "
        f"repump={repump_power_mw:g} mW/beam; "
        f"trapping={1e3*context.trapping_power_w_per_path:.6g} mW/path "
        f"({context.trapping_power_source})",
        flush=True,
    )
    started = perf_counter()
    def progress(completed, total, time_s):
        percent = 100.0 * completed / total
        if progress_widget is not None:
            progress_widget.value = int(round(percent))
            progress_widget.description = f"{percent:4.0f}%"
        print(f"  {percent:5.1f}%  t={1e3*time_s:7.3f} ms", flush=True)
    record = simulate_vector_only_pmot_trajectory(
        launch,
        duration_s=1e-3 * duration_ms,
        context=context,
        trajectory_config=numerical,
        progress_callback=progress,
    )
    wall_time_s = perf_counter() - started
    frame = vector_only_trajectory_dataframe(record)
    capture = record.capture
    final_velocity = np.asarray(record.rate_equation.velocities_m_per_s[-1])
    optical_forces = np.asarray(record.rate_equation.forces_n)
    effective_fields = np.asarray(record.effective_fields_t)
    external_fields = np.asarray(record.rate_equation.magnetic_fields_t)

    tag = output_tag or datetime.now().strftime("run_%Y%m%d_%H%M%S")
    data_dir = DATA_OUTPUT_ROOT / tag
    figure_dir = FIGURE_OUTPUT_ROOT / tag
    csv_path = data_dir / "trajectory.csv"
    metadata_path = data_dir / "run_metadata.json"
    figure_path = figure_dir / "trajectory_and_beams.png"
    if save_outputs:
        save_vector_only_trajectory_csv(record, csv_path)

    title = (
        "Vector-only pMOT: 17 m/s standard launch"
        if radial_distance_mm == 15.0 and speed_m_per_s == 17.0
        else f"Vector-only pMOT: {speed_m_per_s:g} m/s launch"
    )
    figure, axes = plot_pmot_trajectory_diagnostics(
        record,
        list(context.cooling_repump_beams),
        list(context.trapping_beams),
        path=figure_path if save_outputs else None,
        title=title,
        axial_extent_m=1e-3 * plot_extent_mm,
    )
    summary = pd.Series({
        "termination": record.rate_equation.termination_reason,
        "capture classification": capture.classification,
        "trapped by current criterion": capture.trapped,
        "core entries": capture.core_entry_count,
        "maximum continuous core residence [ms]": 1e3 * capture.maximum_continuous_core_residence_s,
        "minimum radius [um]": 1e6 * capture.minimum_radius_m,
        "final radius [um]": 1e6 * capture.final_radius_m,
        "final speed [m/s]": np.linalg.norm(final_velocity),
        "peak optical force [zN]": 1e21 * np.max(np.linalg.norm(optical_forces, axis=1)),
        "peak effective-field proxy [G]": 1e4 * np.max(np.linalg.norm(effective_fields, axis=1)),
        "peak 780 absorption rate [1/s]": frame["total_780_scattering_rate_per_s"].max(),
        "external magnetic field identically zero": bool(np.all(external_fields == 0.0)),
        "trapping power per incident path [mW]": 1e3 * context.trapping_power_w_per_path,
        "saved trajectory samples": len(frame),
        "wall time [s]": wall_time_s,
    }, name="value")
    metadata = {
        "model": record.model_metadata,
        "helicities_propagation_frame": asdict(helicities),
        "launch": {
            "position_m": launch.position_m,
            "velocity_m_per_s": launch.velocity_m_per_s,
            "radial_distance_mm": radial_distance_mm,
            "speed_m_per_s": speed_m_per_s,
            "polar_angle_deg": polar_angle_deg,
            "azimuth_angle_deg": azimuth_angle_deg,
            "impact_parameter_mm": impact_parameter_mm,
            "impact_azimuth_deg": impact_azimuth_deg,
        },
        "integration": {
            "duration_ms": duration_ms,
            "time_step_us": time_step_us,
            "include_diffusion": include_diffusion,
            "include_gravity": include_gravity,
            "seed": int(seed),
            "escape_radius_mm": escape_radius_mm,
        },
        "optics": {
            "cooling_power_mw_per_beam": cooling_power_mw,
            "repump_power_mw_per_beam": repump_power_mw,
            "cooling_detuning_mhz": cooling_detuning_mhz,
            "repump_detuning_mhz": repump_detuning_mhz,
            "beam_diameter_mm": beam_diameter_mm,
            "trapping_wavelength_nm": trapping_wavelength_nm,
            "focus_offset_mm": focus_offset_mm,
            "input_beam_diameter_mm": input_beam_diameter_mm,
            "focal_length_mm": focal_length_mm,
            "retro_power_fraction": retro_power_fraction,
            "trapping_power_w_per_path": context.trapping_power_w_per_path,
            "trapping_power_source": context.trapping_power_source,
            "target_gradient_g_per_cm": context.target_gradient_g_per_cm,
        },
        "capture": asdict(capture),
        "summary": summary.to_dict(),
        "outputs": {
            "trajectory_csv": str(csv_path) if save_outputs else None,
            "metadata_json": str(metadata_path) if save_outputs else None,
            "figure_png": str(figure_path) if save_outputs else None,
        },
    }
    if save_outputs:
        data_dir.mkdir(parents=True, exist_ok=True)
        metadata_path.write_text(
            json.dumps(_json_ready(metadata), indent=2) + "\n",
            encoding="utf-8",
        )
    plt.show()
    display(summary.to_frame())
    if save_outputs:
        print("Saved:")
        print(f"  {csv_path}")
        print(f"  {metadata_path}")
        print(f"  {figure_path}")
    return {
        "record": record,
        "frame": frame,
        "context": context,
        "launch": launch,
        "summary": summary,
        "figure": figure,
        "axes": axes,
        "paths": {
            "csv": csv_path if save_outputs else None,
            "metadata": metadata_path if save_outputs else None,
            "figure": figure_path if save_outputs else None,
        },
    }
```

## Standard 17 m/s trajectory

This cell runs the requested deterministic mean-force trajectory with gravity enabled and recoil diffusion disabled. It writes the complete numerical history, metadata, and figure to the pMOT output tree. Re-run it after changing the keyword arguments if you prefer a script-like workflow; the control panel below is more convenient for exploration.


```python
STANDARD_RUN = run_pmot_case(
    radial_distance_mm=15.0,
    speed_m_per_s=17.0,
    polar_angle_deg=90.0,
    azimuth_angle_deg=0.0,
    impact_parameter_mm=0.0,
    duration_ms=25.0,
    time_step_us=5.0,
    include_diffusion=False,
    include_gravity=True,
    helicities=DEFAULT_HELICITIES,
    save_outputs=True,
    output_tag="standard_17mps",
)
```

## Interactive controls

Changing a control does not launch a calculation. Press **Run pMOT trajectory** after setting the parameters. The six rows below correspond to family and propagation sense; the \(x,y,z\) dropdowns are the helicities/polarizations seen while looking along each component's own propagation vector.

The 35 mm trapping-beam diameter is the collimated beam **before the lens**. The 3D plot instead draws the physical in-trap Gaussian \(1/e^2\) envelopes from the calculated waist and Rayleigh range. Cooling and repump beams occupy the same 12.7 mm diameter Cartesian volumes and are drawn once to avoid a false apparent diameter.


```python
def labeled_number(description, value, *, step=0.1, minimum=None, maximum=None):
    kwargs = dict(description=description, value=value, step=step, continuous_update=False)
    if minimum is not None:
        kwargs["min"] = minimum
    if maximum is not None:
        kwargs["max"] = maximum
    cls = widgets.BoundedFloatText if minimum is not None and maximum is not None else widgets.FloatText
    return cls(**kwargs)

launch_controls = {
    "radial_distance_mm": labeled_number("distance [mm]", 15.0, step=0.5, minimum=0.1, maximum=100.0),
    "speed_m_per_s": labeled_number("speed [m/s]", 17.0, step=0.25, minimum=0.0, maximum=100.0),
    "polar_angle_deg": labeled_number("polar [deg]", 90.0, step=1.0, minimum=0.0, maximum=180.0),
    "azimuth_angle_deg": labeled_number("azimuth [deg]", 0.0, step=1.0, minimum=0.0, maximum=360.0),
    "impact_parameter_mm": labeled_number("impact [mm]", 0.0, step=0.25, minimum=0.0, maximum=50.0),
    "impact_azimuth_deg": labeled_number("impact angle", 0.0, step=1.0, minimum=0.0, maximum=360.0),
}
integration_controls = {
    "duration_ms": labeled_number("duration [ms]", 25.0, step=1.0, minimum=0.01, maximum=500.0),
    "time_step_us": labeled_number("dt [us]", 5.0, step=0.5, minimum=0.1, maximum=100.0),
    "escape_radius_mm": labeled_number("escape [mm]", 30.0, step=1.0, minimum=1.0, maximum=500.0),
    "seed": widgets.BoundedIntText(description="seed", value=20260901, min=0, max=2_147_483_647),
    "include_gravity": widgets.Checkbox(description="gravity", value=True, indent=False),
    "include_diffusion": widgets.Checkbox(description="recoil diffusion", value=False, indent=False),
}
mot_light_controls = {
    "cooling_power_mw": labeled_number("cooling [mW]", 27.0, step=0.5, minimum=0.0, maximum=1000.0),
    "repump_power_mw": labeled_number("repump [mW]", 0.1, step=0.01, minimum=0.0, maximum=100.0),
    "cooling_detuning_mhz": labeled_number("cool det. [MHz]", -15.0, step=0.5),
    "repump_detuning_mhz": labeled_number("rep det. [MHz]", 0.0, step=0.1),
    "beam_diameter_mm": labeled_number("780 diam. [mm]", 12.7, step=0.1, minimum=0.1, maximum=100.0),
}
trapping_controls = {
    "trapping_wavelength_nm": labeled_number("lambda [nm]", 1529.268881, step=0.000001, minimum=1500.0, maximum=1600.0),
    "focus_offset_mm": labeled_number("focus offset", 10.0, step=0.5, minimum=0.01, maximum=100.0),
    "input_beam_diameter_mm": labeled_number("input diam.", 35.0, step=0.5, minimum=0.1, maximum=200.0),
    "focal_length_mm": labeled_number("focal len.", 80.3, step=0.1, minimum=0.1, maximum=1000.0),
    "retro_power_fraction": labeled_number("retro fraction", 1.0, step=0.05, minimum=0.0, maximum=10.0),
    "target_gradient_g_per_cm": labeled_number("gradient proxy", 20.0, step=1.0, minimum=0.01, maximum=1000.0),
    "explicit_trapping_power_mw": labeled_number(
        "trap power [mW]",
        1e3 * STANDARD_RUN["context"].trapping_power_w_per_path,
        step=0.1,
        minimum=0.0,
        maximum=10000.0,
    ),
    "derive_trapping_power": widgets.Checkbox(description="derive from gradient proxy", value=True, indent=False),
    "plot_extent_mm": labeled_number("plot extent", 20.0, step=1.0, minimum=1.0, maximum=100.0),
}

helicity_widgets = {}
grid_items = [
    widgets.HTML("<b>family / path</b>"),
    widgets.HTML("<b>x</b>"),
    widgets.HTML("<b>y</b>"),
    widgets.HTML("<b>z</b>"),
]
for family in ("cooling", "repump", "trapping"):
    for sense in ("incident", "retro"):
        grid_items.append(widgets.HTML(f"<b>{family} {sense}</b>"))
        for index, axis_name in enumerate(AXES):
            options = ("sigma+", "sigma-") if family == "trapping" else ("sigma+", "sigma-", "pi")
            dropdown = widgets.Dropdown(
                options=options,
                value=KNOWN_WORKING_XYZ[index],
                description="",
                layout=widgets.Layout(width="125px"),
            )
            helicity_widgets[(family, sense, axis_name)] = dropdown
            grid_items.append(dropdown)

helicity_grid = widgets.GridBox(
    grid_items,
    layout=widgets.Layout(
        grid_template_columns="170px 130px 130px 130px",
        grid_gap="4px 8px",
        align_items="center",
    ),
)

def control_box(title, controls):
    values = list(controls.values())
    rows = [widgets.HBox(values[index:index+3]) for index in range(0, len(values), 3)]
    return widgets.VBox([widgets.HTML(f"<b>{title}</b>")] + rows)

run_button = widgets.Button(description="Run pMOT trajectory", button_style="primary", icon="play")
save_checkbox = widgets.Checkbox(description="save CSV, JSON, PNG", value=True, indent=False)
progress_bar = widgets.IntProgress(value=0, min=0, max=100, description="ready")
run_output = widgets.Output()

def selected_helicities():
    values = {}
    for family in ("cooling", "repump", "trapping"):
        for sense in ("incident", "retro"):
            values[f"{family}_{sense}_xyz"] = tuple(
                helicity_widgets[(family, sense, axis_name)].value for axis_name in AXES
            )
    return PMOTBeamHelicities(**values)

def widget_values(group):
    return {name: widget.value for name, widget in group.items()}

def run_from_widgets(_=None):
    run_button.disabled = True
    progress_bar.value = 0
    progress_bar.description = "0%"
    progress_bar.bar_style = ""
    with run_output:
        run_output.clear_output(wait=True)
        plt.close("all")
        try:
            stamp = datetime.now().strftime("interactive_%Y%m%d_%H%M%S")
            result = run_pmot_case(
                **widget_values(launch_controls),
                **widget_values(integration_controls),
                **widget_values(mot_light_controls),
                **widget_values(trapping_controls),
                helicities=selected_helicities(),
                save_outputs=save_checkbox.value,
                output_tag=stamp,
                progress_widget=progress_bar,
            )
            globals()["INTERACTIVE_RUN"] = result
            progress_bar.description = "done"
        except Exception:
            progress_bar.bar_style = "danger"
            progress_bar.description = "error"
            raise
        else:
            progress_bar.bar_style = "success"
        finally:
            run_button.disabled = False

run_button.on_click(run_from_widgets)
display(widgets.VBox([
    control_box("Launch geometry", launch_controls),
    control_box("Integration", integration_controls),
    control_box("Cooling and repump", mot_light_controls),
    control_box("1529 nm trapping geometry and power", trapping_controls),
    widgets.HTML("<b>Propagation-frame helicity of every traveling component</b>"),
    helicity_grid,
    widgets.HBox([save_checkbox, run_button, progress_bar]),
    run_output,
]))
```

## Reading the result

The project trapped criterion is met if the atom either stays continuously inside the 2 mm core for at least 5 ms, or enters the core twice with an intervening exit. The summary reports which route was satisfied. Treat this as a diagnostic classification, not yet a validated pMOT capture prediction, until direct trapping-light forces/heating and trajectory convergence checks are included.

The 3D panel uses equal Cartesian limits. Translucent cylinders are the shared cooling/repump volumes at their 12.7 mm diameter. Blue and orange surfaces are the incident and retro 1529 nm Gaussian \(1/e^2\) envelopes, including the \(\pm 10\) mm waist offsets and actual divergence. The tiny diffraction-limited waists are marked with crosses because they are nearly invisible at the full cell scale.

# Atomic And Laser Configuration

This notebook defines the present optical model: collimated cooling and repump beams sharing the same three retroreflected paths. The cooling detuning and repump detuning are explicit user parameters and should be adjusted here before moving on to scattering calculations.

```python
%matplotlib widget
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
SRC_PATH = PROJECT_ROOT / 'src'
if str(SRC_PATH) not in sys.path:
    sys.path.insert(0, str(SRC_PATH))

from pmot import RB87CoolingTransition, RB87RepumpTransition
from pmot.pmot import default_mot_apparatus_config
```

```python
CONFIG = default_mot_apparatus_config()
COOLING_TRANSITION = RB87CoolingTransition()
REPUMP_TRANSITION = RB87RepumpTransition()

beam_rows = [
    {
        'name': CONFIG.cooling.name,
        'role': CONFIG.cooling.role,
        'wavelength_nm': 1e9 * CONFIG.cooling.wavelength_m,
        'beam_diameter_mm': 1e3 * CONFIG.cooling.beam_diameter_m,
        'power_w_per_beam': CONFIG.cooling.power_w_per_beam,
        'resonance_thz': CONFIG.cooling.resonance_frequency_hz / 1e12,
        'detuning_mhz': CONFIG.cooling.detuning_hz / 1e6,
    },
    {
        'name': CONFIG.repump.name,
        'role': CONFIG.repump.role,
        'wavelength_nm': 1e9 * CONFIG.repump.wavelength_m,
        'beam_diameter_mm': 1e3 * CONFIG.repump.beam_diameter_m,
        'power_w_per_beam': CONFIG.repump.power_w_per_beam,
        'resonance_thz': CONFIG.repump.resonance_frequency_hz / 1e12,
        'detuning_mhz': CONFIG.repump.detuning_hz / 1e6,
    },
]
pd.DataFrame(beam_rows)
```

```python
parameter_table = pd.DataFrame([
    {'parameter': 'cooling_beam_diameter_mm', 'value': 1e3 * CONFIG.cooling.beam_diameter_m},
    {'parameter': 'cooling_power_w_per_beam', 'value': CONFIG.cooling.power_w_per_beam},
    {'parameter': 'cooling_detuning_mhz', 'value': CONFIG.cooling.detuning_hz / 1e6},
    {'parameter': 'cooling_resonance_thz', 'value': COOLING_TRANSITION.resonance_frequency_hz / 1e12},
    {'parameter': 'cooling_linewidth_mhz', 'value': COOLING_TRANSITION.linewidth_hz / 1e6},
    {'parameter': 'cooling_saturation_intensity_w_per_m2', 'value': COOLING_TRANSITION.saturation_intensity_w_per_m2},
    {'parameter': 'repump_beam_diameter_mm', 'value': 1e3 * CONFIG.repump.beam_diameter_m},
    {'parameter': 'repump_power_w_per_beam', 'value': CONFIG.repump.power_w_per_beam},
    {'parameter': 'repump_detuning_mhz', 'value': CONFIG.repump.detuning_hz / 1e6},
    {'parameter': 'repump_resonance_thz', 'value': REPUMP_TRANSITION.resonance_frequency_hz / 1e12},
    {'parameter': 'repump_linewidth_mhz', 'value': REPUMP_TRANSITION.linewidth_hz / 1e6},
    {'parameter': 'horizontal_axis_aoi_deg', 'value': CONFIG.axes[0].cell_angle_of_incidence_deg},
    {'parameter': 'beam_axis_names', 'value': tuple(axis.name for axis in CONFIG.axes)},
])
parameter_table
```

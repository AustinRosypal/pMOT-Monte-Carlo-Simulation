# Project Setup

This notebook establishes the project paths, default apparatus configuration, and notebook execution order for the pMOT simulation workflow.

```python
%matplotlib widget
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
SRC_PATH = PROJECT_ROOT / 'src'
if str(SRC_PATH) not in sys.path:
    sys.path.insert(0, str(SRC_PATH))

from pmot.pmot import (
    default_mot_apparatus_config,
    describe_pmot_configuration,
    pmot_notebook_order,
    pmot_paths,
)
```

```python
PATHS = pmot_paths(PROJECT_ROOT)
for path in PATHS.values():
    if isinstance(path, Path) and path.suffix == '':
        path.mkdir(parents=True, exist_ok=True)

CONFIG = default_mot_apparatus_config()
SUMMARY = describe_pmot_configuration(CONFIG)
SUMMARY
```

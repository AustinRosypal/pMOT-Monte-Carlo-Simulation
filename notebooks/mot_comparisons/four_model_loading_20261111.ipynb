# Four-model MOT and pMOT loading overlays

Existing campaign overlays. The fixed controls and acceptance rules differ between models.

All curves use Rb-87 at 296.483 K and 10⁻¹⁰ Torr partial pressure, gravity −z, deterministic mean force.

Coil MOT scans hold 10 mW/cooling beam, −15 MHz, 12.7 mm diameter and axial 20 G/cm except the swept control. Multilevel MOT includes 0.1 mW repump/beam. Their coil gradient ratio is (+G/2,+G/2,−G).

Strange MOT uses 27 mW/cooling beam, −15 MHz, 12.7 mm diameter. Its detuning/diameter scans fix (+2.6113,+2.6113,−5.2226) G/cm. Repump is 0.1 mW/beam, 12.7 mm.

780 nm saturation uses I_sat=16.69 W/m² and linewidth 6.065 MHz. At the fixed −15 MHz, 12.7 mm controls, single-beam (s₀,s_eff) is (9.45968,0.371448) for coil MOTs and (25.5411,1.00291) for strange MOT.

Coil MOTs use a 15 mm-radius disc at 15 mm and a 30 mm escape sphere. Core event is 5 ms inside 2 mm or two entries, with their saved convergence audits. Strange MOT uses a 15 mm disc at 10 mm clipped to a ±14.142 mm cube and requires final 100 ms core residence at 2 s.

Strange trapping light is 1529.268881 nm, equal returns, waists ±10 mm, 35 mm input and 80.3 mm focal length. It supplies a surrogate field to the 24-state MOT kernel.

Bands are pointwise 95% Student-t direction-disc intervals. Curves join simulated settings. The log panel resolves the smaller rates. These overlays do not isolate a single physical change.

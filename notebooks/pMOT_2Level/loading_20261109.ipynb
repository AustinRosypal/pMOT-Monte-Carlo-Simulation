# Effective two-level pMOT loading

Rb-87 effective F=2 → F′=3 mean force. No repumper or real magnetic field. Gravity −z. Six 780.241209686 nm beams, 27 mW each, 12.7 mm 1/e² diameter, −17.5 MHz; matched propagation-frame helicities (σ+, σ+, σ−) on both directions of the 780 and 1529 nm paths. Ideal equal returns, symmetric ±10 mm waist positions, 35 mm input diameter, 80.3 mm focal length. 15 mm launch disc at 10 mm, clipped to ±14.142 mm cube; vapor 296.483 K and 10⁻¹⁰ Torr Rb-87. Bands show pointwise 95% Student-t intervals across independent direction discs. The model includes differential Stark detuning; it omits conservative Stark force, 1529 nm scattering and recoil.

240 settings and 18,960 direction-disc batches. Each setting uses independent directions, with common draws across settings for paired comparisons. Pointwise relative half-width <10%: 0/240 core and 0/240 verified off-core. Zero observed events do not establish zero rate. Replay audited 48 nodes. Seed 2026110914; pilot excluded. Median core relative half-width is 38.4% across 235 settings with observed core retention. The bands describe sampling uncertainty for the fixed 2 s retention rule. Unresolved and unverified localized flux appears separately in each table. Launch points are stratified in impact-parameter annuli with edges 0, 0.2, 0.5, 1, 2, 4, 8, 15 mm and weighted by clipped area.

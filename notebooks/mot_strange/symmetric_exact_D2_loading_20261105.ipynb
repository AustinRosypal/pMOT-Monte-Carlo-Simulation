# Symmetric strange-MOT loading scans

These scans describe a conventional 24-state Rb-87 population-rate MOT driven by an intensity-derived magnetic-field surrogate. The local steady-state solver uses incoherent cooling and repump components, deterministic mean force and gravity.

The 1529 nm light sets the surrogate field geometry and calibration. A full optical pMOT prediction additionally requires a state-resolved Stark Hamiltonian, conservative optical forces and trapping-light scattering and heating.

The gradient axes show the derivative at the origin of the localized surrogate field. They do not describe a spatially uniform coil gradient across the launch aperture. Both trapping-power scans keep the 780 nm cooling and repump settings fixed.

Loading refers to the stated 2 s observation and final 100 ms residence window within the finite cell and launch aperture. Core retention and verified restoring off-core capture are distinct outcomes. Localized trajectories without a verified restoring pocket and other retained trajectories remain separate audit categories. Rates use the declared 0–80 m/s velocity grid, with representative 85–120 m/s tail checks.

Intervals are pointwise 95% Student-t intervals across independent direction discs. Zero observed captures have undefined Student-t precision. Sample counts were fixed before production, and the preferred 10% relative half-width is assessed from the completed sample. These intervals exclude model and velocity-grid errors. Curve maxima describe the sampled settings.

The production audit found 0 verified off-core speed draws and 201,573 localized-only draws (including repeated draws). Off-core verification requires an admissible nearby force balance, stable three-dimensional motion and return after perturbations. A series with no verified draws is omitted from the plots; its zero estimate and undefined interval remain in the tables.

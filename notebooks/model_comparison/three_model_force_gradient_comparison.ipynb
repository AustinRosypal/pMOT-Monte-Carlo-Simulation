# Three-Model MOT Force and Force-Gradient Comparison

This notebook compares the regular 24-state multilevel MOT, `mot_strange` (the same 24-state solver driven by the intensity-derived surrogate field), and the provisional effective two-level pMOT. Every figure uses stationary atoms, excludes gravity, and places force in the top row with the numerical derivative of that same displayed force in the bottom row.

The regular MOT coil and the two-level pMOT vector-shift target are $(+10,+10,-20)$ G/cm. The `mot_strange` field is spatially nonuniform. The two-level pMOT curves are diagnostic rather than validated pMOT predictions.

## 1. Setup and plot generation

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
while PROJECT_ROOT != PROJECT_ROOT.parent:
    if (PROJECT_ROOT / 'pyproject.toml').is_file() and (PROJECT_ROOT / 'src' / 'pmot').is_dir():
        break
    PROJECT_ROOT = PROJECT_ROOT.parent
else:
    raise RuntimeError('Could not locate the pMOT repository root.')

for path in (PROJECT_ROOT, PROJECT_ROOT / 'src'):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

import pandas as pd
from IPython.display import Image, Markdown, display

from scripts.plot_three_model_force_gradient_overlay import run

OUTPUT_DIRECTORY = (
    PROJECT_ROOT
    / 'outputs'
    / 'diagnostics'
    / 'three_model_force_gradient_overlay_20261020'
)
RECOMPUTE_PHYSICS = False
INCLUDE_INDIVIDUAL_MODEL_FIGURES = True

result = run(
    OUTPUT_DIRECTORY,
    reuse_data=not RECOMPUTE_PHYSICS,
    include_individual=INCLUDE_INDIVIDUAL_MODEL_FIGURES,
)
linecuts = pd.read_csv(result['linecuts_csv'])
display(Markdown(f"Saved outputs under `{result['output_directory']}`"))
```

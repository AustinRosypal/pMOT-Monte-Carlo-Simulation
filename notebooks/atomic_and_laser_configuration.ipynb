This notebook collects the experimental parameters used for the beam geometry and field calculations. Values are currently set to the existing apparatus defaults and are intended to be edited here first when the experiment changes.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
SRC_PATH = PROJECT_ROOT / 'src'
if str(SRC_PATH) not in sys.path:
    sys.path.insert(0, str(SRC_PATH))

from pmot import RB87CoolingTransition, default_simulation_config, focused_waist_radius
```

```python
CONFIG = default_simulation_config()
TRANSITION = RB87CoolingTransition()

rows = [{
    'name': 'cooling_780',
    'role': '780 nm cooling overlay',
    'wavelength_nm': 1e9 * CONFIG.cooling.wavelength_m,
    'relative_intensity': 1.0,
    'focused_waist_um': 1e6 * focused_waist_radius(
        wavelength_m=CONFIG.cooling.wavelength_m,
        focal_length_m=CONFIG.lens.focal_length_m,
        input_radius_m=CONFIG.trap_beam.input_radius_m,
    ),
}]

pd.DataFrame(rows)
```

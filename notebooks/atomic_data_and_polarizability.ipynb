# Atomic Data And Polarizability

This notebook loads the Arora CCSD differential-polarizability table, applies the conversion used in the legacy plotting macro, and evaluates the coefficients at the trap wavelengths used in the apparatus.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
SRC_PATH = PROJECT_ROOT / 'src'
if str(SRC_PATH) not in sys.path:
    sys.path.insert(0, str(SRC_PATH))

from pmot import (
    default_polarizability_csv_path,
    default_simulation_config,
    differential_shift_coefficients_for_wavelength,
    plot_polarizability_curves,
    polarizability_dataframe,
    project_paths,
)
```

```python
PATHS = project_paths(PROJECT_ROOT)
CONFIG = default_simulation_config()
CSV_PATH = default_polarizability_csv_path(PROJECT_ROOT)
POLARIZABILITY_DF = polarizability_dataframe(CSV_PATH)
POLARIZABILITY_DF.head()
```

```python
figure_path = PATHS['outputs_figures'] / 'polarizability_conversion.png'
plot_polarizability_curves(POLARIZABILITY_DF, path=figure_path);
```

```python
tone_rows = []
for tone in CONFIG.trap_tones:
    coefficients = differential_shift_coefficients_for_wavelength(1e9 * tone.wavelength_m, CSV_PATH)
    tone_rows.append({
        'tone': tone.name,
        'wavelength_nm': 1e9 * tone.wavelength_m,
        'scalar_mhz_per_intensity': coefficients.scalar_mhz_per_intensity,
        'vector_mhz_per_intensity': coefficients.vector_mhz_per_intensity,
        'tensor_mhz_per_intensity': coefficients.tensor_mhz_per_intensity,
    })

pd.DataFrame(tone_rows)
```
